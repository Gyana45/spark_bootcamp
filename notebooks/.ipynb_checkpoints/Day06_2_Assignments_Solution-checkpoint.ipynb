```python
#question on lead ,lag()
product_data = [
                    (1,"iphone","01-01-2023",1500000),
                    (2,"samsung","01-01-2023",1100000),
                    (3,"oneplus","01-01-2023",1100000),
                    (1,"iphone","01-02-2023",1300000),
                    (2,"samsung","01-02-2023",1120000),
                    (3,"oneplus","01-02-2023",1120000),
                    (1,"iphone","01-03-2023",1600000),
                    (2,"samsung","01-03-2023",1080000),
                    (3,"oneplus","01-03-2023",1160000),
                    (1,"iphone","01-04-2023",1700000),
                    (2,"samsung","01-04-2023",1800000),
                    (3,"oneplus","01-04-2023",1170000),
                    (1,"iphone","01-05-2023",1200000),
                    (2,"samsung","01-05-2023",980000),
                    (3,"oneplus","01-05-2023",1175000),
                    (1,"iphone","01-06-2023",1100000),
                    (2,"samsung","01-06-2023",1100000),
                    (3,"oneplus","01-06-2023",1200000)
                ]
product_schema = StructType(
                            [
                                StructField('product_id',IntegerType(),True),
                                StructField('product_name',StringType(),True),
                                StructField('sale_date',StringType(),True),
                                StructField('sales',IntegerType(),True)
                            ]
                             
                            )



df = spark.createDataFrame(product_data,schema=product_schema)

#convert date_type

df = df.withColumn('sale_date',to_date('sale_date','dd-MM-yyyy'))

df.show(5)
```

# Question -lead/lag

```python
#1.what is the percentage of loss or gain in sales in each product based on previous month sales

window_spec = Window.partitionBy('product_name').orderBy(col('sale_date').asc())

previous_month_df = df.withColumn('previous_month_sales',lag(col('sales'),1).over(window_spec))
previous_month_df.show(5)

previous_month_df.withColumn('%loss_or_gain',round(((col('sales')-col('previous_month_sales'))/col('sales'))*100,2)).show()
```

```python
#2 what is the %age of sale of each month based on last 6months sales

window_spec = Window.partitionBy('product_name').orderBy(col('sale_date').asc())
# if data is not there(NULL),then 0
last_6_months_sales = df.withColumn('last_1m_sales',lag('sales',1,0).over(window_spec))\
                        .withColumn('last_2m_sales',lag('sales',2,0).over(window_spec))\
                        .withColumn('last_3m_sales',lag('sales',3,0).over(window_spec))\
                        .withColumn('last_4m_sales',lag('sales',4,0).over(window_spec))\
                        .withColumn('last_5m_sales',lag('sales',5,0).over(window_spec))\
                        .withColumn('total_6m_sales',col('sales')+col('last_1m_sales')+col('last_2m_sales')+col('last_3m_sales')+col('last_4m_sales')+col('last_5m_sales'))

last_6_months_sales.show(7,truncate=False)

df_final = last_6_months_sales.withColumn('%sales_based_on_last_6m',round((col('sales')/col('total_6m_sales'))*100,2) )\
                                .drop(col('last_1m_sales'),col('last_2m_sales'),col('last_3m_sales'),col('last_4m_sales'),col('last_5m_sales'))\
                                .show()

#-----------using row between----------------

window_spec = Window.partitionBy('product_id')\
                    .orderBy('sale_date')\
                    .rowsBetween(-5,0)

df_result = df.withColumn('total_6m_sales',sum('sales').over(window_spec))\
                .withColumn('pct_sales',round((col('sales')/col('total_6m_sales'))*100,2))\
                .show()
```

```python
orders_schema_struct = StructType(
                                    [
                                        StructField("order_id",IntegerType()),
                                        StructField("customer_id",IntegerType()),
                                        StructField("product_id",IntegerType()),
                                        StructField("price",FloatType()),
                                        StructField("order_date",DateType()),
                                        StructField("order_status",StringType()),
                                        StructField("state",StringType()),
                                        StructField("quantity",IntegerType()),
                                    ]
                                )

df_orders = spark.read.csv('/data/orders_50mb.csv',schema=orders_schema_struct,header=True)

df_orders.show(5)

df_orders.createOrReplaceTempView('orders_table')
```

```python
#product table
product_schema_struct = StructType(
                                    [
                                    StructField('product_id',IntegerType()),
                                    StructField('product_name',StringType()),
                                    StructField('cost_price',FloatType())
                                    ]
                                )

df_product = spark.read.csv('/data/products.csv',schema=product_schema_struct,header=True)

df_product.show(5)

df_product.createOrReplaceTempView('product_table')
```

# Question-1

1. Using the orders and products datasets, find the **top 2 most profitable products in each state**, where profit is defined as `(price − cost_price) * quantity`. Consider only delivered orders. Return state, product_id, total profit, and rank within the state.


```python
df1 =df_orders.filter(lower(col('order_status'))=='delivered')\
            .join(df_product,'product_id','inner')\
            .withColumn('profit',
                        (coalesce(col('price'),lit(0)) - coalesce(col('cost_price'),lit(0)))*coalesce(col('quantity'),lit(0))
                       )\
            .groupBy(col('state'),col('product_id'))\
                .agg(
                        sum(coalesce(col('profit'),lit(0))).alias('total_profit')
                    )


window_spec = Window.partitionBy(col('state')).orderBy(col('total_profit').desc())
df_final = df1.withColumn('dense_rank',dense_rank().over(window_spec))\
                .filter(col('dense_rank')<=2)
df_final.show()
            
```

```python
#spark.sql

spark.sql("""
            with agg_cte AS
                (
                    select o.state,
                            p.product_name,
                            sum((o.price-p.cost_price)*o.quantity) as product_profit
                    from orders_table o
                    join(product_table p)
                    on o.product_id = p.product_id
                    where o.order_status='DELIVERED'
                    group by state,product_name
                ),
            ranked_cte AS
                (select *,dense_rank() over(partition by state order by product_profit desc) as rank
                from agg_cte
                )
            select * from ranked_cte where rank<=2
            
           


"""
         ).show()

```

# Question 2

2. For each customer, identify their **most recent order** and calculate the **difference between 
that order’s price and the average price of all products purchased by that customer**.
Return customer_id, order_id, order_date, price, and the price difference.


```python
window_spec = Window.partitionBy('customer_id').orderBy(col('order_date').desc())
window_spec_2 = Window.partitionBy('customer_id') #if i will go with window_spec for avg,then it will give avg according to order_date desc

df = df_orders.withColumn('row_num',row_number().over(window_spec))\
            .withColumn('avg_price',avg(col('price')).over(window_spec_2))\
            .filter(col('row_num')==1)\
            .withColumn('price_diff',col('price')-col('avg_price'))\
            .select('customer_id','order_id','order_date','price','avg_price','price_diff')\
            .orderBy(col("customer_id").asc())
print(df.count())
df.show()
```

```python
window_latest = Window.partitionBy("customer_id").orderBy(col("order_date").desc())

df_latest = df_orders.withColumn(
    "rn",
    row_number().over(window_latest)
).filter(col("rn") == 1)

# Step 2: Average unit price per customer
df_avg = df_orders.groupBy("customer_id").agg(avg("price").alias("avg_price"))

# Step 3: Join latest order with avg price
df_final = df_latest.join(df_avg, "customer_id", "inner")

# Step 4: Calculate difference
df_result = df_final.withColumn("price_diff",col("price") - col("avg_price"))

# Final output
df_result = df_result.select(
                                "customer_id",
                                "order_id",
                                "order_date",
                                col("price").alias("unit_price"),
                                "avg_price",
                                "price_diff"
                            ).orderBy(col("customer_id").asc())

print(df_result.count())
df_result.show(truncate=False)
```

```python
spark.sql(
            """
            with
                latest_cte AS 
                    (select *,
                        row_number() over(partition by customer_id order by order_date desc) as rn
                        from orders_table),
                agg_cte AS
                     (select customer_id,avg(price) as avg_price from latest_cte 
                         group by customer_id
                    )
                select l.customer_id,l.order_id,l.price,a.avg_price ,(l.price-a.avg_price) as price_diff
                from latest_cte l 
                inner join agg_cte a
                on l.customer_id=a.customer_id
                where rn =1
                order by l.customer_id asc

            """
).show()
```

# Question-3

3.Using orders and products data, determine for each year the **product that contributed the highest percentage of total yearly revenue**. Return order_year, product_id, product_revenue, total_year_revenue, and revenue_percentage.

```python
df1= df_orders.withColumn('revenue',col('price')*col('quantity'))\
                .withColumn('order_year',year(col('order_date')))

df_year_product_revenue = df1.groupBy(col('order_year'),col('product_id'))\
                                .agg(
                                        sum(col('revenue')).alias('product_revenue')
                                )

df_year_revenue = df1.groupBy(col('order_year'))\
                        .agg(
                                sum(col('revenue')).alias('total_year_revenue')
                        )

df_join = df_year_product_revenue.join(df_year_revenue,'order_year')

df = df_join.withColumn('revenue_percentage',round((col('product_revenue')/col('total_year_revenue'))*100,2))

window_spec = Window.partitionBy('order_year').orderBy(col('revenue_percentage').desc())
df.withColumn('rank',rank().over(window_spec))\
    .filter(col('rank')==1)\
    .orderBy(col('order_year').asc())\
    .show()

```

```python
df_orders.withColumn('prod_price',col('price')*col('quantity'))\
            .join(df_product,'product_id','inner')\
            .groupBy('product_id','product_name',year(col('order_date')).alias('year'))\
                .agg(sum('prod_price').alias('year_prod_revenue'))\
            .withColumn('year_total_revenue',sum('year_prod_revenue').over(Window.partitionBy(col('year'))))\
            .withColumn('revenue_percentage',(col('year_prod_revenue')/col('year_total_revenue'))*100)\
            .withColumn('Dense_rank',dense_rank().over(Window.partitionBy('year').orderBy(col('revenue_percentage').desc())))\
            .orderBy(col('year').asc(),col('Dense_rank').asc(),col('product_id').asc())\
            .filter(col('Dense_rank')==1)\
            .show()
```

# Question 4

4. Using the orders and products datasets, calculate for each state the running cumulative revenue over time,
ordered by order_date, 
and return state, order_date, order_id, order_value, and cumulative revenue.

```python
#orderid are not in ordered,so writing it in last
window_spec =Window.partitionBy('state').orderBy(col('order_date').asc(),col('order_id').asc())
df_orders.withColumn('order_value',col('price')*col('quantity'))\
            .withColumn('running_price',sum('order_value').over(window_spec).alias('running_price'))\
            .select('state','order_date','order_id','order_value','running_price')\
            .orderBy(col('state').asc(),col('order_date').asc(),col('order_id').asc())\
            .show()
```

```python
# Step 1: Join orders with products (optional, not required for revenue)
df_joined = df_orders.join(df_product, "product_id", "inner")

# Step 2: Calculate order value
df_value = df_joined.withColumn("order_value",col("price") * col("quantity"))

# Step 3: Define window (state-wise, ordered by date)
window_spec = Window.partitionBy("state").orderBy("order_date",'order_id').rowsBetween(Window.unboundedPreceding, Window.currentRow)

# Step 4: Calculate cumulative revenue
df_result = df_value.withColumn("cumulative_revenue",sum("order_value").over(window_spec))

# Final output
df_result = df_result.select("state","order_date","order_id","order_value","cumulative_revenue")\
                        .orderBy("state", "order_date",'order_id')


df_result.show(truncate=False)
```

# Question-5
5.For each product, find the first and last order date along with the total revenue generated between those dates, considering only non cancelled orders.


```python
df_orders.filter(lower(col('order_status'))!='cancelled')\
            .join(df_product,'product_id','inner').drop(col('cost_price'))\
            .withColumn('revenue',col('price')*col('quantity'))\
            .groupBy('product_id','product_name')\
                .agg(
                        min('order_date').alias('first_order_date'),
                        max('order_date').alias('last_order_date'),
                        sum(col('revenue')).alias('total_revenue')
                )\
            .show(5)
```

# Question -6
6. Identify customers whose latest order value is greater than their previous order value, and return customer_id, order_id, order_date, previous_order_value, and latest_order_value.


```python

window_spec = Window.partitionBy('customer_id').orderBy(col('order_date').desc())

df_orders.withColumn('current_revenue',col('price')*col('quantity'))\
            .withColumn('prev_revenue',lead(col('current_revenue'),1).over(window_spec))\
            .withColumn('row_num',row_number().over(window_spec))\
            .filter(col('row_num')==1)\
            .filter(col('prev_revenue').isNotNull())\
            .filter((col('current_revenue')-col('prev_revenue'))>0)\
            .select('customer_id','order_id','order_date','current_revenue','prev_revenue')\
            .orderBy(col('customer_id').asc())\
            .show()
```

# Question-7
7.Using orders and products data, rank products within each year based on total profit, and return only the top 3 products per year.


```python
# df_orders.filter(col('product_id')==70).show(3)
# df_product.filter(col('product_id')==70).show(3)
df_orders.join(df_product,'product_id','inner')\
            .withColumn('order_year',year(col('order_date')))\
            .withColumn('order_profit',round((col('price')-col('cost_price'))*col('quantity'),2))\
            .groupBy('order_year','product_id','product_name')\
                .agg(
                        sum(col('order_profit')).alias('product_total_profit')
                )\
            .withColumn('dense_rank',dense_rank().over(Window.partitionBy(col('order_year')).orderBy(col('product_total_profit').desc())))\
            .filter(col('dense_rank')<=3)\
            .withColumn('product_total_profit',round(col('product_total_profit'),2))\
            .orderBy(col('order_year').asc(),col('dense_rank').asc())\
            .show()
```
